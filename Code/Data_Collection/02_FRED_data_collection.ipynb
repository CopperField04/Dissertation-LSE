# 02 -- FRED Data Collection

## Source
Federal Reserve Economic Data (FRED) via the `fredapi` Python library. API key authenticated.

## Collection Method
Two pull methods are used depending on whether the series is subject to data revisions:

- **Standard pull** (`fred.get_series`): Used for market and financial data that is not revised after initial publication (e.g., Treasury yields, credit spreads, commodity prices).
- **First release pull** (`fred.get_series_first_release`): Used for macro data that undergoes revisions (e.g., employment, inflation, GDP components). This retrieves the vintage value as it was first published, preventing look-ahead bias from revised figures. Falls back to standard pull if first-release retrieval fails.

All series are pulled from 2003-01-01 to 2024-12-31 with a 0.2-second pause between API calls.

## Derived Factors

### Daily Derived
- **Yield curve slopes:** `slope_2y10y`, `slope_3m10y`, `slope_2y30y`, `slope_1y10y`
- **Yield curve curvature (butterfly):** `curve_2_5_10` (2x5Y - 2Y - 10Y), `curve_2_10_30` (2x10Y - 2Y - 30Y)
- **Credit spread differentials:** `bbb_aaa_spread`, `bb_bbb_spread`, `moody_baa_aaa`
- **Real rates:** `real_rate_10y` (10Y nominal - 10Y breakeven), `real_rate_5y`
- **Policy rate spreads:** `ff_2y_spread` (2Y - Fed Funds), `ff_10y_spread` (10Y - Fed Funds)
- **Commodity ratios:** `oil_gold_ratio` (WTI / Gold), `brent_wti_spread` (Brent - WTI)
- **Equity returns:** `sp500_ret_1d` (1-day), `sp500_ret_5d` (5-day) percentage changes

### Monthly Derived
- **Month-over-month % changes** (`_mom` suffix) for 26 level series including nonfarm payrolls, CPI, PCE, industrial production, retail sales, housing starts, durable orders, money supply, and copper
- **Year-over-year % changes** (`_yoy` suffix) for: `cpi_urban`, `cpi_core`, `pce_price`, `pce_core`, `ppi_final`, `case_shiller`
- **Copper/gold ratio:** monthly copper price divided by monthly-averaged daily gold price
- **Trade balance 12-month moving average:** `trade_balance_12m_avg`

## Pipeline Notes (Documented in Notebook)
- Daily: merge onto trading calendar, forward-fill gaps up to 5 days
- Weekly: forward-fill to daily, then merge onto trading calendar; compute `net_liquidity = fed_assets - tga - rrp` downstream
- Monthly (first release): two-month publication lag applied before use (e.g., January data available from first trading day of March)
- Monthly (standard -- money supply, commodity prices): one-month lag for consistency

## Filtering
Daily output filtered to 2004-01-01 through 2024-12-31. Weekly and monthly outputs filtered from 2003-01-01 through 2024-12-31.

```python
# %% [markdown]
# # FRED Data Collection — Comprehensive
#
# Pulls all available FRED series relevant to US equity market prediction.
# Two pull methods:
#
# - **Standard pull** (`get_series`): Market/financial data with no revisions.
# - **First release pull** (`get_series_first_release`): Macro data that gets revised.
#
# Outputs:
# - `data/fred_daily.parquet` — daily-frequency factors
# - `data/fred_weekly.parquet` — weekly-frequency factors
# - `data/fred_monthly_macro.parquet` — monthly macro factors (first-release)

# %%
import pandas as pd
import numpy as np
from fredapi import Fred
from pathlib import Path
import time


FRED_API_KEY = 'XXXXXXXXX'  # ← Replace with your key
fred = Fred(api_key=FRED_API_KEY)

START = '2003-01-01'
END = '2024-12-31'
PAUSE = 0.2  # seconds between API calls

# %%
def pull_standard(series_dict, start=START, end=END):
    """Pull multiple FRED series using standard (no-revision) method."""
    frames = {}
    for col, sid in series_dict.items():
        try:
            s = fred.get_series(sid, observation_start=start, observation_end=end)
            frames[col] = s
            n = s.dropna().shape[0]
            d0 = s.dropna().index.min().date() if n > 0 else 'N/A'
            d1 = s.dropna().index.max().date() if n > 0 else 'N/A'
            print(f"  ✓ {col:<35s} ({sid:<20s}): {n:>5d} obs | {d0} → {d1}")
        except Exception as e:
            print(f"  ✗ {col:<35s} ({sid:<20s}): FAILED — {e}")
        time.sleep(PAUSE)
    return frames

def pull_first_release(series_dict, start=START, end=END):
    """Pull multiple FRED series using first-release (vintage) method.
    Falls back to standard pull if first_release not available."""
    frames = {}
    for col, sid in series_dict.items():
        try:
            s = fred.get_series_first_release(sid)
            s = s[(s.index >= start) & (s.index <= end)]
            frames[col] = s
            n = s.dropna().shape[0]
            d0 = s.dropna().index.min().date() if n > 0 else 'N/A'
            d1 = s.dropna().index.max().date() if n > 0 else 'N/A'
            print(f"  ✓ {col:<35s} ({sid:<20s}) [1st REL]: {n:>5d} obs | {d0} → {d1}")
        except Exception as e:
            print(f"  ⚠ {col:<35s} ({sid:<20s}): first_release failed, trying standard...")
            try:
                s = fred.get_series(sid, observation_start=start, observation_end=end)
                frames[col] = s
                n = s.dropna().shape[0]
                print(f"    → Standard fallback: {n} obs")
            except Exception as e2:
                print(f"    → Also failed: {e2}")
        time.sleep(PAUSE)
    return frames

def frames_to_df(frames, date_start='2004-01-01', date_end='2024-12-31'):
    """Combine series dict into a single DataFrame, filtered to date range."""
    df = pd.DataFrame(frames)
    df.index.name = 'date'
    df = df.reset_index()
    df['date'] = pd.to_datetime(df['date'])
    df = df[(df['date'] >= date_start) & (df['date'] <= date_end)].reset_index(drop=True)
    return df

# %% [markdown]
# ---
# # 1. Daily Market/Financial Data (Standard Pull — no revisions)

# %% [markdown]
# ## 1a. Treasury Yields — Full Curve

# %%
print("=" * 80)
print("TREASURY YIELDS — FULL CURVE")
print("=" * 80)

treasury_yields = {
    'yield_1m':     'DGS1MO',       # 1-Month Treasury
    'yield_3m':     'DTB3',          # 3-Month Treasury Bill
    'yield_6m':     'DTB6',          # 6-Month Treasury Bill
    'yield_1y':     'DGS1',          # 1-Year Treasury
    'yield_2y':     'DGS2',          # 2-Year Treasury
    'yield_3y':     'DGS3',          # 3-Year Treasury
    'yield_5y':     'DGS5',          # 5-Year Treasury
    'yield_7y':     'DGS7',          # 7-Year Treasury
    'yield_10y':    'DGS10',         # 10-Year Treasury
    'yield_20y':    'DGS20',         # 20-Year Treasury
    'yield_30y':    'DGS30',         # 30-Year Treasury
}

treasury_frames = pull_standard(treasury_yields)

# %% [markdown]
# ## 1b. TIPS Yields & Breakeven Inflation

# %%
print("\n" + "=" * 80)
print("TIPS YIELDS & BREAKEVEN INFLATION")
print("=" * 80)

tips_breakeven = {
    'tips_5y':          'DFII5',         # 5-Year TIPS Real Yield
    'tips_7y':          'DFII7',         # 7-Year TIPS Real Yield
    'tips_10y':         'DFII10',        # 10-Year TIPS Real Yield
    'tips_20y':         'DFII20',        # 20-Year TIPS Real Yield
    'tips_30y':         'DFII30',        # 30-Year TIPS Real Yield
    'breakeven_5y':     'T5YIE',         # 5-Year Breakeven Inflation
    'breakeven_7y':     'T7YIE',         # 7-Year Breakeven Inflation (may not exist)
    'breakeven_10y':    'T10YIE',        # 10-Year Breakeven Inflation
    'breakeven_20y':    'T20YIEM',       # 20-Year Breakeven (monthly — check)
    'breakeven_30y':    'T30YIEM',       # 30-Year Breakeven (monthly — check)
}

tips_frames = pull_standard(tips_breakeven)

# %% [markdown]
# ## 1c. Policy Rates & Interbank

# %%
print("\n" + "=" * 80)
print("POLICY RATES & INTERBANK")
print("=" * 80)

policy_rates = {
    'fed_funds_eff':    'DFF',           # Effective Federal Funds Rate
    'sofr':             'SOFR',          # Secured Overnight Financing Rate
    'prime_rate':       'DPRIME',        # Bank Prime Loan Rate
    'discount_rate':    'DPCREDIT',      # Discount Window Primary Credit Rate
    'libor_3m':         'USD3MTD156N',   # 3-Month LIBOR (discontinued 2023)
    'iorb':             'IORB',          # Interest on Reserve Balances
}

policy_frames = pull_standard(policy_rates)

# %% [markdown]
# ## 1d. Credit Spreads — Comprehensive

# %%
print("\n" + "=" * 80)
print("CREDIT SPREADS")
print("=" * 80)

credit_spreads = {
    # ICE BofA Option-Adjusted Spreads
    'hy_oas':           'BAMLH0A0HYM2',     # US High Yield OAS
    'ig_oas':           'BAMLC0A0CM',        # US Investment Grade OAS
    'aaa_oas':          'BAMLC0A1CAAA',      # AAA OAS
    'aa_oas':           'BAMLC0A2CAA',       # AA OAS
    'a_oas':            'BAMLC0A3CA',        # A OAS
    'bbb_oas':          'BAMLC0A4CBBB',      # BBB OAS
    'bb_oas':           'BAMLH0A1HYBB',      # BB OAS
    'b_oas':            'BAMLH0A2HYB',       # B OAS
    'ccc_oas':          'BAMLH0A3HYC',       # CCC & below OAS

    # Moody's seasoned corporate
    'moody_aaa':        'DAAA',              # Moody's AAA Corporate Bond Yield
    'moody_baa':        'DBAA',              # Moody's BAA Corporate Bond Yield

    # TED spread
    'ted_spread':       'TEDRATE',           # TED Spread (ended ~2022)
}

credit_frames = pull_standard(credit_spreads)

# %% [markdown]
# ## 1e. Mortgage & Consumer Rates

# %%
print("\n" + "=" * 80)
print("MORTGAGE & CONSUMER RATES")
print("=" * 80)

mortgage_rates = {
    'mortgage_30y':     'MORTGAGE30US',      # 30-Year Fixed Mortgage Rate (weekly)
    'mortgage_15y':     'MORTGAGE15US',      # 15-Year Fixed Mortgage Rate (weekly)
}

mortgage_frames = pull_standard(mortgage_rates)

# %% [markdown]
# ## 1f. Commodities

# %%
print("\n" + "=" * 80)
print("COMMODITIES")
print("=" * 80)

commodities = {
    'wti_oil':          'DCOILWTICO',        # WTI Crude Oil ($/barrel)
    'brent_oil':        'DCOILBRENTEU',      # Brent Crude Oil ($/barrel)
    'natgas':           'DHHNGSP',           # Henry Hub Natural Gas ($/MMBtu)
    'gold':             'GOLDAMGBD228NLBM',  # Gold London PM Fix ($/oz)
}

commodity_frames = pull_standard(commodities)

# %% [markdown]
# ## 1g. Dollar Index & Financial Conditions

# %%
print("\n" + "=" * 80)
print("DOLLAR INDEX & FINANCIAL CONDITIONS")
print("=" * 80)

dollar_conditions = {
    # Trade-weighted dollar
    'twexb':            'DTWEXBGS',          # Broad Trade-Weighted Dollar Index
    'twexm':            'DTWEXEMEGS',        # EM Trade-Weighted Dollar Index

    # Fed Financial Conditions (weekly, but FRED may serve daily-aligned)
    'nfci':             'NFCI',              # Chicago Fed National Financial Conditions
    'anfci':            'ANFCI',             # Adjusted NFCI
    'stlfsi':           'STLFSI2',           # St. Louis Fed Financial Stress Index

    # CBOE
    'move_index':       'MOVE',              # MOVE Index (bond vol) — may not exist on FRED
}

dollar_frames = pull_standard(dollar_conditions)

# %% [markdown]
# ## 1h. Fed Liquidity — Daily Component

# %%
print("\n" + "=" * 80)
print("FED LIQUIDITY — DAILY")
print("=" * 80)

fed_daily = {
    'rrp':              'RRPONTSYD',         # Overnight Reverse Repo ($B)
    'soma_treasury':    'TREAST',            # Fed Treasury Holdings ($M) — weekly
    'soma_mbs':         'MBST',              # Fed MBS Holdings ($M) — weekly (may fail)
}

fed_daily_frames = pull_standard(fed_daily)

# %% [markdown]
# ## 1i. Equity Market Indicators

# %%
print("\n" + "=" * 80)
print("EQUITY MARKET INDICATORS")
print("=" * 80)

equity_indicators = {
    'sp500':            'SP500',             # S&P 500 Index Level
    'wilshire5000':     'WILL5000IND',       # Wilshire 5000 Total Market
    'margin_debt':      'BOGZ1FL663067003Q', # Margin debt (quarterly — check)
}

equity_frames = pull_standard(equity_indicators)

# %% [markdown]
# ---
# # Combine all daily frames

# %%
all_daily_frames = {}
for d in [treasury_frames, tips_frames, policy_frames, credit_frames,
          mortgage_frames, commodity_frames, dollar_frames, fed_daily_frames,
          equity_frames]:
    all_daily_frames.update(d)

fred_daily = frames_to_df(all_daily_frames)

print(f"\n{'=' * 80}")
print(f"COMBINED DAILY DATA: {fred_daily.shape[0]} rows, "
      f"{fred_daily.shape[1] - 1} factors")
print(f"{'=' * 80}")

# %% [markdown]
# ## Compute derived daily factors

# %%
# === YIELD CURVE SHAPE ===
# Slopes
if 'yield_10y' in fred_daily and 'yield_2y' in fred_daily:
    fred_daily['slope_2y10y'] = fred_daily['yield_10y'] - fred_daily['yield_2y']
if 'yield_10y' in fred_daily and 'yield_3m' in fred_daily:
    fred_daily['slope_3m10y'] = fred_daily['yield_10y'] - fred_daily['yield_3m']
if 'yield_30y' in fred_daily and 'yield_2y' in fred_daily:
    fred_daily['slope_2y30y'] = fred_daily['yield_30y'] - fred_daily['yield_2y']
if 'yield_10y' in fred_daily and 'yield_1y' in fred_daily:
    fred_daily['slope_1y10y'] = fred_daily['yield_10y'] - fred_daily['yield_1y']

# Curvature (butterfly): 2×middle − short − long
if all(c in fred_daily for c in ['yield_2y', 'yield_5y', 'yield_10y']):
    fred_daily['curve_2_5_10'] = (
        2 * fred_daily['yield_5y'] - fred_daily['yield_2y'] - fred_daily['yield_10y']
    )
if all(c in fred_daily for c in ['yield_2y', 'yield_10y', 'yield_30y']):
    fred_daily['curve_2_10_30'] = (
        2 * fred_daily['yield_10y'] - fred_daily['yield_2y'] - fred_daily['yield_30y']
    )

# === CREDIT SPREADS ===
if 'bbb_oas' in fred_daily and 'aaa_oas' in fred_daily:
    fred_daily['bbb_aaa_spread'] = fred_daily['bbb_oas'] - fred_daily['aaa_oas']
if 'bb_oas' in fred_daily and 'bbb_oas' in fred_daily:
    fred_daily['bb_bbb_spread'] = fred_daily['bb_oas'] - fred_daily['bbb_oas']
if 'moody_baa' in fred_daily and 'moody_aaa' in fred_daily:
    fred_daily['moody_baa_aaa'] = fred_daily['moody_baa'] - fred_daily['moody_aaa']

# === REAL RATES ===
if 'yield_10y' in fred_daily and 'breakeven_10y' in fred_daily:
    fred_daily['real_rate_10y'] = fred_daily['yield_10y'] - fred_daily['breakeven_10y']
if 'yield_5y' in fred_daily and 'breakeven_5y' in fred_daily:
    fred_daily['real_rate_5y'] = fred_daily['yield_5y'] - fred_daily['breakeven_5y']

# === POLICY RATE SPREADS ===
if 'fed_funds_eff' in fred_daily and 'yield_2y' in fred_daily:
    fred_daily['ff_2y_spread'] = fred_daily['yield_2y'] - fred_daily['fed_funds_eff']
if 'fed_funds_eff' in fred_daily and 'yield_10y' in fred_daily:
    fred_daily['ff_10y_spread'] = fred_daily['yield_10y'] - fred_daily['fed_funds_eff']

# === COMMODITY RATIOS ===
if 'wti_oil' in fred_daily and 'gold' in fred_daily:
    fred_daily['oil_gold_ratio'] = fred_daily['wti_oil'] / fred_daily['gold']
if 'brent_oil' in fred_daily and 'wti_oil' in fred_daily:
    fred_daily['brent_wti_spread'] = fred_daily['brent_oil'] - fred_daily['wti_oil']

# === EQUITY ===
if 'sp500' in fred_daily:
    fred_daily['sp500_ret_1d'] = fred_daily['sp500'].pct_change()
    fred_daily['sp500_ret_5d'] = fred_daily['sp500'].pct_change(5)

print(f"After derived factors: {fred_daily.shape[1] - 1} total factors")
print(f"New derived columns: {[c for c in fred_daily.columns if c not in all_daily_frames and c != 'date']}")

# %% [markdown]
# ---
# # 2. Weekly Data (Standard Pull — no revisions)

# %%
print("\n" + "=" * 80)
print("WEEKLY MARKET DATA (standard pull)")
print("=" * 80)

weekly_market = {
    # Fed Balance Sheet
    'fed_assets':       'WALCL',             # Fed Total Assets ($M)
    'tga':              'WTREGEN',           # Treasury General Account ($M)
    'reserves':         'WRESBAL',           # Reserve Balances at Fed ($M)

    # Banking
    'bank_credit':      'TOTBKCR',           # Commercial Bank Credit ($B)
    'ci_loans':         'TOTCI',             # Commercial & Industrial Loans ($B)
    'consumer_loans':   'CONSUMER',          # Consumer Loans ($B)
    'real_estate_loans':'REALLN',            # Real Estate Loans ($B)
}

weekly_market_frames = pull_standard(weekly_market)

# %% [markdown]
# ## Weekly Macro Data (First Release)

# %%
print("\n" + "=" * 80)
print("WEEKLY MACRO DATA (first release)")
print("=" * 80)

weekly_macro = {
    'initial_claims':   'ICSA',              # Initial Jobless Claims
    'continued_claims': 'CCSA',              # Continued Jobless Claims
}

weekly_macro_frames = pull_first_release(weekly_macro)

# %%
# Combine all weekly frames
all_weekly_frames = {}
all_weekly_frames.update(weekly_market_frames)
all_weekly_frames.update(weekly_macro_frames)

fred_weekly = frames_to_df(all_weekly_frames, date_start='2003-01-01')

print(f"\n{'=' * 80}")
print(f"COMBINED WEEKLY DATA: {fred_weekly.shape[0]} rows, "
      f"{fred_weekly.shape[1] - 1} factors")
print(f"{'=' * 80}")

# %% [markdown]
# ---
# # 3. Monthly Macro Data (First Release)
#
# Two-month lag in pipeline: data for January published in February,
# model only sees it from March onwards.

# %% [markdown]
# ## 3a. Employment & Labor

# %%
print("=" * 80)
print("MONTHLY MACRO — EMPLOYMENT & LABOR (first release)")
print("=" * 80)

employment = {
    'nonfarm_payrolls':     'PAYEMS',        # Total Nonfarm Payrolls (thousands)
    'unrate':               'UNRATE',        # Unemployment Rate (%)
    'u6_rate':              'U6RATE',        # U-6 Underemployment Rate (%)
    'participation':        'CIVPART',       # Labor Force Participation Rate (%)
    'avg_hourly_earnings':  'CES0500000003', # Avg Hourly Earnings ($/hr)
    'avg_weekly_hours':     'AWHMAN',        # Avg Weekly Hours, Manufacturing
    'jolts_openings':       'JTSJOL',        # JOLTS Job Openings (thousands)
    'jolts_quits':          'JTSQUR',        # JOLTS Quits Rate (%)
}

employment_frames = pull_first_release(employment)

# %% [markdown]
# ## 3b. Inflation

# %%
print("\n" + "=" * 80)
print("MONTHLY MACRO — INFLATION (first release)")
print("=" * 80)

inflation = {
    'cpi_urban':            'CPIAUCSL',      # CPI-U All Items (SA)
    'cpi_core':             'CPILFESL',      # CPI-U Less Food & Energy (SA)
    'cpi_food':             'CPIUFDSL',      # CPI-U Food (SA)
    'cpi_energy':           'CPIENGSL',      # CPI-U Energy (SA)
    'cpi_shelter':          'CUSR0000SAH1',  # CPI-U Shelter
    'cpi_services':         'CUSR0000SAS',   # CPI-U Services
    'pce_price':            'PCEPI',         # PCE Price Index
    'pce_core':             'PCEPILFE',      # Core PCE Price Index
    'ppi_final':            'PPIFIS',        # PPI Final Demand
    'ppi_core':             'PPIFES',        # PPI Final Demand Less Foods/Energy
    'import_prices':        'IR',            # Import Price Index
}

inflation_frames = pull_first_release(inflation)

# %% [markdown]
# ## 3c. Production & Output

# %%
print("\n" + "=" * 80)
print("MONTHLY MACRO — PRODUCTION & OUTPUT (first release)")
print("=" * 80)

production = {
    'indpro':               'INDPRO',        # Industrial Production Index
    'cap_util':             'TCU',           # Total Capacity Utilization (%)
    'manuf_prod':           'IPMAN',         # Manufacturing Production Index
    'durable_orders':       'DGORDER',       # Durable Goods Orders ($M)
    'durable_ex_transport': 'ADXTNO',        # Durable Goods ex Transportation ($M)
    'factory_orders':       'AMTMNO',        # Factory Orders ($M)
    'business_inventories': 'BUSINV',        # Total Business Inventories ($M)
    'inv_sales_ratio':      'ISRATIO',       # Inventory/Sales Ratio
}

production_frames = pull_first_release(production)

# %% [markdown]
# ## 3d. Consumer & Spending

# %%
print("\n" + "=" * 80)
print("MONTHLY MACRO — CONSUMER & SPENDING (first release)")
print("=" * 80)

consumer = {
    'retail_sales':         'RSAFS',         # Retail Sales ($M, SA)
    'retail_ex_auto':       'RSFSXMV',       # Retail Sales ex Motor Vehicles ($M)
    'personal_income':      'PI',            # Personal Income ($B, SAAR)
    'personal_spending':    'PCE',           # Personal Consumption Expenditure ($B)
    'saving_rate':          'PSAVERT',       # Personal Saving Rate (%)
    'consumer_credit':      'TOTALSL',       # Total Consumer Credit ($B)
    'umich_sentiment':      'UMCSENT',       # U of Michigan Consumer Sentiment
    'conf_board':           'CSCICP03USM665S', # OECD Consumer Confidence (proxy)
    'vehicle_sales':        'TOTALSA',       # Total Vehicle Sales (M, SAAR)
}

consumer_frames = pull_first_release(consumer)

# %% [markdown]
# ## 3e. Housing

# %%
print("\n" + "=" * 80)
print("MONTHLY MACRO — HOUSING (first release)")
print("=" * 80)

housing = {
    'housing_starts':       'HOUST',         # Housing Starts (thousands, SAAR)
    'building_permits':     'PERMIT',        # Building Permits (thousands, SAAR)
    'new_home_sales':       'HSN1F',         # New Home Sales (thousands, SAAR)
    'existing_home_sales':  'EXHOSLUSM495S', # Existing Home Sales (M, SA)
    'case_shiller':         'CSUSHPINSA',    # S&P/CS US National Home Price Index
    'nahb_index':           'USHMI',         # NAHB Housing Market Index (may fail)
}

housing_frames = pull_first_release(housing)

# %% [markdown]
# ## 3f. Trade

# %%
print("\n" + "=" * 80)
print("MONTHLY MACRO — TRADE (first release)")
print("=" * 80)

trade = {
    'trade_balance':        'BOPGSTB',       # Trade Balance ($M)
    'exports':              'BOPGEXP',       # Exports ($M)
    'imports':              'BOPGIMP',       # Imports ($M)
}

trade_frames = pull_first_release(trade)

# %% [markdown]
# ## 3g. Money Supply & Velocity

# %%
print("\n" + "=" * 80)
print("MONTHLY MACRO — MONEY SUPPLY (standard — no revisions)")
print("=" * 80)

money_supply = {
    'm1':                   'M1SL',          # M1 Money Stock ($B, SA)
    'm2':                   'M2SL',          # M2 Money Stock ($B, SA)
    'm2_velocity':          'M2V',           # Velocity of M2 (quarterly)
    'monetary_base':        'BOGMBASE',      # Monetary Base ($M)
}

money_frames = pull_standard(money_supply)

# %% [markdown]
# ## 3h. Commodities (Monthly)

# %%
print("\n" + "=" * 80)
print("MONTHLY COMMODITIES (standard — market prices, no revisions)")
print("=" * 80)

monthly_commodities = {
    'copper_monthly':       'PCOPPUSDM',     # Copper ($/metric ton)
    'aluminum_monthly':     'PALUMUSDM',     # Aluminum ($/metric ton)
    'wheat_monthly':        'PWHEAMTUSDM',   # Wheat ($/metric ton)
    'corn_monthly':         'PMAIZMTUSDM',   # Corn/Maize ($/metric ton)
    'lumber_monthly':       'WPU0811',       # PPI Lumber (proxy — may work)
}

monthly_commodity_frames = pull_standard(monthly_commodities)

# %% [markdown]
# ## 3i. ISM / PMI Proxies & Business Surveys

# %%
print("\n" + "=" * 80)
print("MONTHLY — ISM / BUSINESS SURVEYS (first release)")
print("=" * 80)

business_surveys = {
    'ism_manuf':            'MANEMP',        # ISM Manufacturing Employment
    'ism_prices':           'PPIFIS',        # ISM Prices proxy (PPI)
    'ism_new_orders':       'NEWORDER',      # ISM New Orders (if available)
    'philly_fed':           'GACDFSA066MSFRBPHI',  # Philly Fed Manufacturing
    'empire_state':         'GACDISA066MSFRBNY',   # Empire State Manufacturing
    'kansas_fed':           'KCFSI',         # KC Fed Financial Stress
    'chicago_fed_nai':      'CFNAI',         # Chicago Fed National Activity Index
    'lei':                  'USSLIND',       # Leading Economic Index (Conference Board)
}

business_survey_frames = pull_first_release(business_surveys)

# %% [markdown]
# ---
# # Combine all monthly frames

# %%
all_monthly_frames = {}
for d in [employment_frames, inflation_frames, production_frames,
          consumer_frames, housing_frames, trade_frames, money_frames,
          monthly_commodity_frames, business_survey_frames]:
    all_monthly_frames.update(d)

fred_monthly = frames_to_df(all_monthly_frames, date_start='2003-01-01')

print(f"\n{'=' * 80}")
print(f"COMBINED MONTHLY DATA: {fred_monthly.shape[0]} rows, "
      f"{fred_monthly.shape[1] - 1} factors")
print(f"{'=' * 80}")

# %% [markdown]
# ## Compute derived monthly factors

# %%
# Month-over-month percent changes for level series
mom_cols = [
    'nonfarm_payrolls', 'cpi_urban', 'cpi_core', 'pce_price', 'pce_core',
    'ppi_final', 'ppi_core', 'indpro', 'retail_sales', 'retail_ex_auto',
    'personal_income', 'personal_spending', 'housing_starts',
    'building_permits', 'new_home_sales', 'existing_home_sales',
    'durable_orders', 'durable_ex_transport', 'factory_orders',
    'exports', 'imports', 'm1', 'm2', 'monetary_base',
    'copper_monthly', 'case_shiller', 'consumer_credit',
]

for col in mom_cols:
    if col in fred_monthly.columns:
        fred_monthly[f'{col}_mom'] = fred_monthly[col].pct_change() * 100

# Year-over-year for key inflation measures
yoy_cols = ['cpi_urban', 'cpi_core', 'pce_price', 'pce_core',
            'ppi_final', 'case_shiller']
for col in yoy_cols:
    if col in fred_monthly.columns:
        fred_monthly[f'{col}_yoy'] = fred_monthly[col].pct_change(12) * 100

# Ratios
if 'copper_monthly' in fred_monthly.columns:
    # Copper/gold ratio needs monthly gold — average daily gold to monthly
    if 'gold' in all_daily_frames:
        gold_s = pd.Series(all_daily_frames['gold'])
        gold_monthly = gold_s.resample('MS').mean()
        gold_monthly = gold_monthly[(gold_monthly.index >= '2003-01-01') &
                                     (gold_monthly.index <= '2024-12-31')]
        # Align with fred_monthly dates
        copper_gold = []
        for _, row in fred_monthly.iterrows():
            d = row['date']
            # Find closest monthly gold
            closest = gold_monthly.index[gold_monthly.index <= d]
            if len(closest) > 0:
                g = gold_monthly.loc[closest[-1]]
                c = row.get('copper_monthly', np.nan)
                copper_gold.append(c / g if pd.notna(c) and pd.notna(g) and g > 0
                                   else np.nan)
            else:
                copper_gold.append(np.nan)
        fred_monthly['copper_gold_ratio'] = copper_gold

# Trade balance as % of trend
if 'trade_balance' in fred_monthly.columns:
    fred_monthly['trade_balance_12m_avg'] = (
        fred_monthly['trade_balance'].rolling(12).mean()
    )

n_derived = len([c for c in fred_monthly.columns
                 if c.endswith('_mom') or c.endswith('_yoy') or
                 c.endswith('_ratio') or c.endswith('_avg')])
print(f"\nAdded {n_derived} derived monthly factors")
print(f"Total monthly factors: {fred_monthly.shape[1] - 1}")

# %% [markdown]
# ---
# # 4. NaN Summary

# %%
def nan_summary(df, name):
    """Print concise NaN summary for a DataFrame."""
    print(f"\n{'=' * 80}")
    print(f"{name}: {df.shape[0]} rows, {df.shape[1] - 1} factors")
    print(f"{'=' * 80}")
    cols = [c for c in df.columns if c != 'date']
    for c in sorted(cols):
        n = df[c].isna().sum()
        pct = n / len(df) * 100
        if pct > 0:
            valid = df[df[c].notna()]['date']
            last = valid.max().date() if len(valid) > 0 else 'N/A'
            first = valid.min().date() if len(valid) > 0 else 'N/A'
            flag = " ← INVESTIGATE" if pct > 30 else ""
            print(f"  {c:<35s}: {n:>5d} NaN ({pct:>5.1f}%) "
                  f"| {first} → {last}{flag}")
        else:
            print(f"  {c:<35s}: no NaN ✓")

nan_summary(fred_daily, "DAILY")
nan_summary(fred_weekly, "WEEKLY")
nan_summary(fred_monthly, "MONTHLY MACRO")

# %% [markdown]
# ---
# # 5. Save Outputs

# %%
fred_daily.to_parquet('../../Data/Data_Collection/Initial/02_FRED/fred_daily.parquet', index=False, engine='pyarrow')
print(f"Saved Data/Data_Collection/Initial/02_FRED/fred_daily.parquet: {fred_daily.shape}")

fred_weekly.to_parquet('../../Data/Data_Collection/Initial/02_FRED/fred_weekly.parquet', index=False, engine='pyarrow')
print(f"Saved Data/Data_Collection/Initial/02_FRED/fred_weekly.parquet: {fred_weekly.shape}")

fred_monthly.to_parquet('../../Data/Data_Collection/Initial/02_FRED/fred_monthly_macro.parquet',
                         index=False, engine='pyarrow')
print(f"Saved Data/Data_Collection/Initial/02_FRED/fred_monthly_macro.parquet: {fred_monthly.shape}")

# %% [markdown]
# ---
# # 6. Complete Factor List

# %%
print("=" * 80)
print("COMPLETE FACTOR INVENTORY")
print("=" * 80)

print(f"\n--- DAILY ({fred_daily.shape[1] - 1} factors) ---")
for i, c in enumerate([x for x in fred_daily.columns if x != 'date'], 1):
    print(f"  {i:>3d}. {c}")

print(f"\n--- WEEKLY ({fred_weekly.shape[1] - 1} factors) ---")
for i, c in enumerate([x for x in fred_weekly.columns if x != 'date'], 1):
    print(f"  {i:>3d}. {c}")

print(f"\n--- MONTHLY ({fred_monthly.shape[1] - 1} factors) ---")
for i, c in enumerate([x for x in fred_monthly.columns if x != 'date'], 1):
    print(f"  {i:>3d}. {c}")

total = (fred_daily.shape[1] - 1 + fred_weekly.shape[1] - 1 +
         fred_monthly.shape[1] - 1)
print(f"\n{'=' * 80}")
print(f"TOTAL FRED FACTORS: {total}")
print(f"{'=' * 80}")

print("""
PIPELINE NOTES:
• Daily: Merge onto trading calendar. Forward-fill gaps ≤5 days.
• Weekly: Forward-fill to daily, then merge onto trading calendar.
  Compute net_liquidity = fed_assets - tga - rrp in pipeline.
• Monthly (first release): TWO-month lag before use.
  Factor dated January → available from first trading day of March.
• Monthly (standard): Also two-month lag for consistency, except
  money supply and commodity prices which use one-month lag.
""")

print("\nFRED collection complete.")
```
